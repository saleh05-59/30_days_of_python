```python
# triangle area calculator script
triangle_base = float(input("Enter base: "))
triangle_height = float(input("Enter height: "))
triangle_area = (triangle_base * triangle_height) / 2
print(f"The area of the triangle is {triangle_area}")
```

```python
# triangle perimeter calculator cript
triangle_side_a = float(input("Enter side a: "))
triangle_side_b = float(input("Enter side b: "))
triangle_side_c = float(input("Enter side c: "))
triangle_perimeter = triangle_side_a + triangle_side_b + triangle_side_c
print(f"The perimeter of the triangle is {triangle_perimeter}")
```

```python
# rectangle perimeter and area calculator
rectangle_length = float(input("Enter length: "))
rectangle_width = float(input("Enter width: "))
rectangle_area = rectangle_length * rectangle_width
rectangle_perimeter = (rectangle_length + rectangle_width) * 2
print(f"the area of rectangle is {rectangle_area} and the rectangle's perimeter is {rectangle_perimeter}")
```

```python
# circle perimeter and area calculator
circle_radius = float(input("Enter radius: "))
circle_area = 3.14 * (circle_radius ** 2)
circle_perimeter = 2 * 3.14 * circle_radius
print(f"circle's area is {circle_area} and it's perimeter is {circle_perimeter}")
```

```python
# y = 2x - 2
m = 2
b = -2 # == y-intercept

# for x-intercept we need y = 0
x_intercept = (0 - b) / m

print(f"slope = {m}")
print(f"y-intercept = {b}")
print(f"x-intercept = {x_intercept}")
```

```python

#  Euclidean distance between (2, 2) and (6,10)
x1, y1 = 2, 2
x2, y2 = 6, 10

m_task_9 = y2-y1/x2-x1 # slope formula
print(f"the euclidean distance is {m_task_9}")
```

```python
# comparing the two slopes
print(f"the second slope is much more steeper than the first one {m_task_9} > {m}")
```

```python
# y = x^2 + 6x + 9
def question_formula(x):
    y = (x**2) + (6*x) + 9
    return y

x_test = [1, 2, 3, 0, -1, -2, -3]
for i in x_test :
    y = question_formula(i)
    if y != 0 :
        print(f"x = {i} , y = {y}")
    else :
        print(f"huzzahh! {i} makes y = 0")
```

```python
# pay calculator
hours_work = float(input("Enter hours: "))
pay_rate_work = float(input("Enter rate per hour: "))
weekly_earning = hours_work * pay_rate_work
print(f"Your weekly earning is {weekly_earning}")
```

```python
# life calculator
age = float(input("Enter your age: "))
expected_die = 100
years_to_live = expected_die - age
sec_to_live = years_to_live * 365 * 24 * 60 * 60
print(f"You have {sec_to_live} seconds to die")
```
